```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(0)
X = torch.tensor(
    [
        [1. , -1.] ,
        [0.,2.],
        [9.,-8.],
        [7.,1.]
    ]
)
```

```python
y = torch.tensor([1, 0, 1, 0])
```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()

        self.fcl = nn.Linear(2,4)
        self.relu  = nn.ReLU()

        self.fc2 = nn.Linear(4,2)

    def forward(self ,x):
      x=self.fcl(x)
      x=self.relu(x)
      x=self.fc2(x)

      return x
```

```python
model = MLP()
```

```python
output = model(X)
```

```python
print(output)
```

```python
loss_function = nn.CrossEntropyLoss()
```

```python
loss = loss_function(output, y)
```

```python
loss
```

```python
loss.backward()
```

```python
loss
```

```python
learning_rate = 0.1
with torch.no_grad():
    for parameters in model.parameters():
      parameters -= learning_rate*parameters.grad
```

```python
learning_rate = 0.1
```

```python
output = model(X)
```

```python
loss = loss_function(output,y)
```

```python
loss
```

```python
epochs = 20
for epoch in range(epochs):
  output = model(X)
  model.zero_grad()
  loss = loss_function(output,y)
  loss.backward()
  with torch.no_grad():
    for parameters in model.parameters():
      parameters -= learning_rate*parameters.grad

  print(
    "Epoch , Loss " , epoch , loss.item()
 )
```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()

        self.fcl = nn.Linear(2,4)
        self.sigmoid1  = nn.Sigmoid()

        self.fc2 = nn.Linear(4,2)
        self.sigmoid2 = nn.Sigmoid()
        self.fc3 = nn.Linear(2,2)

    def forward(self ,x):
      x=self.fcl(x)
      x=self.sigmoid(x)
      x=self.fc2(x)
      x=self.sigmoid(x)
      x=self.fc3(x)

      return x
```

```python
model = MLP()
```

```python
model.sigmoid = model.sigmoid1
output = model(X)
```

```python
print(output)
```

```python
loss_function = nn.CrossEntropyLoss()
```

```python
loss = loss_function(output, y)
```

```python
loss
```
