```python
import numpy as np


class Node:
    def __init__(self, prediction):
        self.prediction = prediction

        self.feature = None
        self.threshold = None
        self.missing_left = None

        self.left = None
        self.right = None


class CART:
    def __init__(self, max_depth=3, min_samples_leaf=3):
        self.max_depth = max_depth
        self.min_samples_leaf = min_samples_leaf
        self.root = None
        self.root_candidates = []

    # ---------------------------------------------------------
    # GINI IMPURITY
    # ---------------------------------------------------------
    def gini(self, y):
        if len(y) == 0:
            return 0.0

        _, counts = np.unique(y, return_counts=True)

        probabilities = counts / len(y)

        return 1.0 - np.sum(probabilities ** 2)

    # ---------------------------------------------------------
    # MAJORITY CLASS
    # ---------------------------------------------------------
    def majority_class(self, y):
        values, counts = np.unique(y, return_counts=True)

        # np.unique sorts values, so np.argmax gives
        # deterministic smallest-label tie breaking.
        return values[np.argmax(counts)]

    # ---------------------------------------------------------
    # GENERATE LEGAL THRESHOLDS
    # ---------------------------------------------------------
    def get_thresholds(self, x):
        finite_values = x[np.isfinite(x)]

        if len(finite_values) < 2:
            return []

        values = np.unique(finite_values)

        if len(values) < 2:
            return []

        return (values[:-1] + values[1:]) / 2.0

    # ---------------------------------------------------------
    # EVALUATE ONE SPLIT
    # ---------------------------------------------------------
    def evaluate_split(self, X, y, indices, feature, threshold,
                       missing_left):

        x = X[indices, feature]

        missing = np.isnan(x)

        left = (~missing) & (x <= threshold)
        right = (~missing) & (x > threshold)

        if missing_left:
            left = left | missing
        else:
            right = right | missing

        left_indices = indices[left]
        right_indices = indices[right]

        # Minimum child size constraint
        if (len(left_indices) < self.min_samples_leaf or
                len(right_indices) < self.min_samples_leaf):
            return None

        parent_gini = self.gini(y[indices])

        left_gini = self.gini(y[left_indices])
        right_gini = self.gini(y[right_indices])

        n = len(indices)

        weighted_child_gini = (
            (len(left_indices) / n) * left_gini
            +
            (len(right_indices) / n) * right_gini
        )

        gain = parent_gini - weighted_child_gini

        return {
            "feature": feature,
            "threshold": float(threshold),
            "missing_left": missing_left,
            "left_indices": left_indices,
            "right_indices": right_indices,
            "n_left": len(left_indices),
            "n_right": len(right_indices),
            "gini_left": left_gini,
            "gini_right": right_gini,
            "gain": gain
        }

    # ---------------------------------------------------------
    # FIND ALL CANDIDATE SPLITS
    # ---------------------------------------------------------
    def find_candidates(self, X, y, indices):

        candidates = []

        n_features = X.shape[1]

        for feature in range(n_features):

            x = X[indices, feature]

            thresholds = self.get_thresholds(x)

            for threshold in thresholds:

                # Missing -> left
                result = self.evaluate_split(
                    X,
                    y,
                    indices,
                    feature,
                    threshold,
                    True
                )

                if result is not None:
                    candidates.append(result)

                # Missing -> right
                result = self.evaluate_split(
                    X,
                    y,
                    indices,
                    feature,
                    threshold,
                    False
                )

                if result is not None:
                    candidates.append(result)

        return candidates

    # ---------------------------------------------------------
    # TIE-BREAKING KEY
    # ---------------------------------------------------------
    def candidate_sort_key(self, candidate):

        # Python sorts ascending.
        #
        # We want:
        # 1. larger gain
        # 2. smaller feature
        # 3. smaller threshold
        # 4. missing-left before missing-right
        #
        return (
            -candidate["gain"],
            candidate["feature"],
            candidate["threshold"],
            0 if candidate["missing_left"] else 1
        )

    # ---------------------------------------------------------
    # BUILD TREE
    # ---------------------------------------------------------
    def build(self, X, y, indices, depth):

        prediction = self.majority_class(y[indices])

        node = Node(prediction)

        # Stop if maximum depth reached
        if depth >= self.max_depth:
            return node

        # Stop if node is pure
        if self.gini(y[indices]) == 0.0:
            return node

        candidates = self.find_candidates(X, y, indices)

        if not candidates:
            return node

        candidates.sort(key=self.candidate_sort_key)

        best = candidates[0]

        node.feature = best["feature"]
        node.threshold = best["threshold"]
        node.missing_left = best["missing_left"]

        node.left = self.build(
            X,
            y,
            best["left_indices"],
            depth + 1
        )

        node.right = self.build(
            X,
            y,
            best["right_indices"],
            depth + 1
        )

        return node

    # ---------------------------------------------------------
    # FIT
    # ---------------------------------------------------------
    def fit(self, X, y):

        X = np.asarray(X, dtype=float)
        y = np.asarray(y)

        indices = np.arange(len(y))

        # Save every root candidate for Q1 reporting
        self.root_candidates = self.find_candidates(
            X, y, indices
        )

        self.root_candidates.sort(
            key=self.candidate_sort_key
        )

        self.root = self.build(
            X,
            y,
            indices,
            depth=0
        )

        return self

    # ---------------------------------------------------------
    # PREDICT ONE OBSERVATION
    # ---------------------------------------------------------
    def predict_one(self, x):

        node = self.root

        while node.feature is not None:

            value = x[node.feature]

            if np.isnan(value):

                if node.missing_left:
                    node = node.left
                else:
                    node = node.right

            elif value <= node.threshold:
                node = node.left

            else:
                node = node.right

        return node.prediction

    # ---------------------------------------------------------
    # PREDICT DATASET
    # ---------------------------------------------------------
    def predict(self, X):

        X = np.asarray(X, dtype=float)

        return np.array([
            self.predict_one(row)
            for row in X
        ])
```
