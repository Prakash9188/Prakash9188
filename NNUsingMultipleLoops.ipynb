```python

import math

# Step 1: Training data for the AND gate
X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

Y = [0, 0, 0, 1]

# Step 2: Define network architecture
input_size = 2
hidden_size = 2
output_size = 1

# Step 3: Initialize every weight to the same value
W1 = [[0.5 for j in range(hidden_size)]
      for i in range(input_size)]

W2 = [[0.5 for j in range(output_size)]
      for i in range(hidden_size)]

b1 = [0.0 for j in range(hidden_size)]
b2 = [0.0 for k in range(output_size)]

learning_rate = 0.5
epochs = 10000


# Step 4: Activation function
def sigmoid(z):
    return 1 / (1 + math.exp(-z))


# Derivative of sigmoid, using its output
def sigmoid_derivative(a):
    return a * (1 - a)


# Step 5: Training
for epoch in range(epochs):

    total_loss = 0

    for sample in range(len(X)):

        # ----- FORWARD PROPAGATION -----

        # Input layer to hidden layer
        hidden = [0.0 for j in range(hidden_size)]

        for j in range(hidden_size):
            z = b1[j]

            for i in range(input_size):
                z += X[sample][i] * W1[i][j]

            hidden[j] = sigmoid(z)

        # Hidden layer to output layer
        output = [0.0 for k in range(output_size)]

        for k in range(output_size):
            z = b2[k]

            for j in range(hidden_size):
                z += hidden[j] * W2[j][k]

            output[k] = sigmoid(z)

        # ----- CALCULATE ERROR -----

        target = Y[sample]
        prediction = output[0]

        # Binary cross-entropy loss
        p = max(1e-10, min(1 - 1e-10, prediction))
        total_loss += -(
            target * math.log(p)
            + (1 - target) * math.log(1 - p)
        )

        # ----- BACKPROPAGATION -----

        # Output delta: sigmoid + cross-entropy
        delta_out = [0.0 for k in range(output_size)]

        for k in range(output_size):
            delta_out[k] = output[k] - target

        # Hidden-layer delta
        delta_hidden = [0.0 for j in range(hidden_size)]

        for j in range(hidden_size):
            error = 0.0

            for k in range(output_size):
                error += W2[j][k] * delta_out[k]

            delta_hidden[j] = (
                error * sigmoid_derivative(hidden[j])
            )

        # ----- UPDATE OUTPUT WEIGHTS -----

        for j in range(hidden_size):
            for k in range(output_size):
                W2[j][k] -= (
                    learning_rate * hidden[j] * delta_out[k]
                )

        for k in range(output_size):
            b2[k] -= learning_rate * delta_out[k]

        # ----- UPDATE INPUT WEIGHTS -----

        for i in range(input_size):
            for j in range(hidden_size):
                W1[i][j] -= (
                    learning_rate
                    * X[sample][i]
                    * delta_hidden[j]
                )

        for j in range(hidden_size):
            b1[j] -= learning_rate * delta_hidden[j]

    if epoch % 1000 == 0:
        print(
            "Epoch:", epoch,
            "Loss:", round(total_loss / len(X), 6)
        )


# Step 6: Test the trained network
print("\nFinal predictions:")

for sample in range(len(X)):

    hidden = [0.0 for j in range(hidden_size)]

    for j in range(hidden_size):
        z = b1[j]

        for i in range(input_size):
            z += X[sample][i] * W1[i][j]

        hidden[j] = sigmoid(z)

    z = b2[0]

    for j in range(hidden_size):
        z += hidden[j] * W2[j][0]

    prediction = sigmoid(z)

    print(
        X[sample],
        "Probability:", round(prediction, 4),
        "Predicted class:", int(prediction >= 0.5),
        "Actual:", Y[sample]
    )
```
