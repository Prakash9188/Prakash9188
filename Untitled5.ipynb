```python
import numpy as np

#Create Node

class Node:
    def __init__(self, prediction):
      self.prediction = prediction
      self.feature = None
      self.threshold = None
      self.missing_left = None
      self.missing_right = None
      self.left = None
      self.right = None

#Cart

class Cart:
  def __init__(self , max_depth = 3 , min_samples_leaf = 3):
    self.max_depth = max_depth
    self.min_sample_leaf = min_samples_leaf
    self.root = None
    self.root_candidates = []
```

```python
#Gini impurity

def gini(self, y):
  if len(y) == 0 :
    return 0.0

  _, counts = np.unique(y , return_counts = True)
  probabilities = counts / len(y)

  return 1.0 - np.sum(probabilities ** 2)

#Majority

def majority_class(slef , y ) :
  values , counts = np.unique(y , return_counts = True)
  return values[np.argmax(counts)]

#Finding Legal Threshold

def get_thresholds(self, values):
    finite_values = values[np.isfinite(values)]
    unique_values = np.unique(finite_values)

    if len(unique_values) < 2:
        return np.array([], dtype=float)

    thresholds = (unique_values[:-1] + unique_values[1:]) / 2.0
    return thresholds
```

```python
#Split

def evaluate_split(
        self,
        X,
        y,
        indices,
        feature,
        threshold,
        missing_left
    ):
        values = X[
            indices,
            feature
        ]

        missing = np.isnan(values)

        left_mask = (
            (~missing)& (values <= threshold)
        )

        right_mask = (
            (~missing) & (values > threshold)
        )

  # Decide where missing values go

        if missing_left:
            left_mask = left_mask | missing
        else:
            right_mask = right_mask | missing

 # Location of child

        left_indices = indices[left_mask]
        right_indices = indices[right_mask]

 #Min leaf

        if len(left_indices) < self.min_samples_leaf:
            return None

        if len(right_indices) < self.min_samples_leaf:
            return None
# Parent Gini

        parent_gini = self.gini(
            y[indices]
        )

# Child Gini

        left_gini = self.gini(y[left_indices])

        right_gini = self.gini(y[right_indices])

        n = len(indices)

# Weighted child Gini

        weighted_child_gini = (
            (len(left_indices) / n) * left_gini
            +
            (len(right_indices) / n) * right_gini
        )

# Gain

        gain = (
            parent_gini
            -
            weighted_child_gini
        )

        return {
            "feature": feature,
            "threshold": float(threshold),
            "missing_left": missing_left,

            "left_indices": left_indices,
            "right_indices": right_indices,

            "n_left": len(left_indices),
            "n_right": len(right_indices),

            "gini_left": left_gini,
            "gini_right": right_gini,

            "gain": gain
        }
```

```python
# All Candidate Split

def find_candidates(self , X , y , indices):
  candidates = []

  n_feature = X.shape[1]
  for feature in range(n_feature) :
    values = X[indices , feature]
    thresholds = self.get_thresholds(values)

# left
    for threshold in thresholds :
      candidate = self.evaluate_split(
          X , y , indices ,
          feature , threshold , True
      )

      if candidate is not None :
        candidates.append(candidate)

#right
      candidate = self.evaluate_split(
          X ,y , indices ,
          feature , threshold , False
      )

      if candidate is not None :
        candidates.append(candidate)
  return candidates
```

```python
#sorting

def sort_key(self , candidate):
  return(
      -candidate["gain"],
      candidate["features"],
      candidate["threshold"],
      0 if candidate["missing_left"] else 1
  )
```

```python
# Building the Tree

def build_tree(
        self,
        X,
        y,
        indices,
        depth
    ):
    prediction = self.majority_class(y[indices])

    node = Node(prediction)
    if depth >= self.max_depth :
      return node

    if self.gini(y[indices]) == 0.0 :
      return node

    candidates = self.find_candidates(X , y , indices)

    if len(candidates) == 0 :
      return node

# best candidate

    candidates.sort(key = self.sort_key)
    best = candidates[0]
    node.feature = best["feature"]
    node.threshold = best["threshold"]
    node.missing_left = best["missing_left"]

    node.left = self.build_tree(
        X , y , best["left_indices"] , depth + 1 )
    node.right = self.build_tree(
        X , y , best["right_indices"] , depth + 1 )

    return node
```

```python
#Prediction for ONe Obs
def predict_one(self , x ) :
  node = self.root
  while node.feature is not None :
    value = x[node.feature]
    if np.isnan(value):
      node = node.left if node.missing_left else node.right

    elif value <= node.threshold :
      node = node.left

    else :
      node = node.right
    return node.prediction
```
